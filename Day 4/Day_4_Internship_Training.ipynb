To visualize the distribution of numerical features, we first need to identify them and calculate their skewness. Then we can use the `sns.histplot` to plot them.

```python
numerical_features = Data.select_dtypes(include=['float64', 'int64'])
skewness = numerical_features.skew()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 10))
for i, column in enumerate(numerical_features.columns, 1):
    plt.subplot((len(numerical_features.columns) // 3) + 1, 3, i)
    sns.histplot(numerical_features[column].dropna(), kde=True, bins=30, color='blue')
    plt.title(f'{column} (Skewness: {skewness[column]:.2f})', fontsize=10)
    plt.xlabel(column, fontsize=8)
    plt.ylabel('Frequency', fontsize=8)

plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

encoded_data = company_data.copy()

# 1. Label Encoding for 'Sex'
label_encoder= LabelEncoder()
encoded_data['Gender_Label'] = label_encoder.fit_transform(encoded_data['Gender'])

# 2. One-Hot Encoding for 'Embarked'
one_hot_encoded = pd.get_dummies(encoded_data['Department'], prefix='Department')
encoded_data = pd.concat([encoded_data, one_hot_encoded], axis=1)


# Display the resulting dataset with relevant encoded columns
print(encoded_data[['Gender', 'Gender_Label', 'Department','Department_IT','Department_HR','Department_Marketing','Department_Finance','Department_Sales' ]].head())
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
company_data=pd.read_csv("/content/preprocessing_practice_dataset.csv")

# Select numerical columns for scaling
numerical_columns = ['Age', 'Monthly_Salary', 'Department']

# Drop rows with missing values for simplicity
company_data.dropna(subset=numerical_columns, inplace=True)

# Initialize scalers
minmax_scaler = MinMaxScaler()
standard_scaler = StandardScaler()

# Apply Min-Max Scaling
company_data['Age_MinMax'] = minmax_scaler.fit_transform(company_data[['Age']])
company_data['Salary_MinMax'] = minmax_scaler.fit_transform(company_data[['Monthly_Salary']])

# Apply Standard Scaling
company_data['Age_Standard'] = standard_scaler.fit_transform(company_data[['Age']])
company_data['Salary_Standard'] = standard_scaler.fit_transform(company_data[['Monthly_Salary']])


# Display scaled features
company_data[['Age', 'Age_MinMax', 'Age_Standard','Salary_MinMax','Salary_Standard']].head()
```
