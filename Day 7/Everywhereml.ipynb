```python
from sklearn.datasets import load_iris
from everywhereml.sklearn.tree import DecisionTreeClassifier

# Load the Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# Create and train a DecisionTreeClassifier
# Using max_depth=3 for better visualization in the Arduino code
clf_iris = DecisionTreeClassifier(max_depth=3)
clf_iris.fit(X, y)

# Print the Arduino representation of the trained model
print(clf_iris.to_arduino())
```
